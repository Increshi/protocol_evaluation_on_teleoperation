# WebRTC Throughput Analysis Notebook
This notebook visualizes the throughput (send and receive rates) of WebRTC DataChannels using the `webrtc_channel_stats_*.csv` file generated by Unity.

```python
import pandas as pd
import matplotlib.pyplot as plt
import os
```

## Analysis Function
Define the function to plot the throughput metrics.

```python
def plot_throughput_analysis(stats_csv_file, output_filename="throughput_analysis.png"):
    """
    Plots the throughput (Send and Receive) for WebRTC DataChannels.
    Expected CSV format: webrtc_channel_stats_*.csv
    """
    if not os.path.exists(stats_csv_file):
        print(f"❌ Could not find {stats_csv_file}")
        print("Please ensure you specify the correct path to the stats CSV.")
        return

    # Load data
    df = pd.read_csv(stats_csv_file)
    
    # Convert timestamp to datetime for better plotting
    df['timestamp_utc'] = pd.to_datetime(df['timestamp_utc'])
    
    # Filter by channel
    df_robot = df[df['channel_label'] == 'robot_state']
    df_cmd = df[df['channel_label'] == 'commands']

    # Setup the plot
    fig, axes = plt.subplots(2, 1, figsize=(10, 8), sharex=True)

    # --- Plot 1: Robot State Throughput ---
    axes[0].plot(df_robot['timestamp_utc'], df_robot['throughput_send_kbps'], label="Robot State (Send)", color='salmon', linestyle='--')
    axes[0].plot(df_robot['timestamp_utc'], df_robot['throughput_recv_kbps'], label="Robot State (Receive)", color='blue', linewidth=2)
    
    axes[0].set_title('Robot State Channel (Unreliable) Throughput')
    axes[0].set_ylabel('Throughput (kbps)')
    axes[0].legend(loc="upper left")
    axes[0].grid(True, linestyle=':', alpha=0.6)

    # --- Plot 2: Commands Throughput ---
    axes[1].plot(df_cmd['timestamp_utc'], df_cmd['throughput_send_kbps'], label="Commands (Send)", color='red', linewidth=2)
    axes[1].plot(df_cmd['timestamp_utc'], df_cmd['throughput_recv_kbps'], label="Commands (Receive)", color='lightblue', linestyle='--')
    
    axes[1].set_title('Commands Channel (Reliable) Throughput')
    axes[1].set_xlabel('Time (UTC)')
    axes[1].set_ylabel('Throughput (kbps)')
    axes[1].legend(loc="upper left")
    axes[1].grid(True, linestyle=':', alpha=0.6)

    # Finalize and Save
    fig.autofmt_xdate() # Auto-rotate timestamp labels
    plt.tight_layout()
    plt.savefig(output_filename, dpi=300)
    print(f"✅ Saved Throughput Analytics plot to: {output_filename}")
    
    # Display inline in the notebook
    plt.show()
```

## Execute Analysis
Set your actual CSV file name and run the code below to visualize.

```python
# Execute the analysis (Replace with your actual stats CSV filename)
stats_csv = "webrtc_channel_stats_YOUR_TIMESTAMP.csv"
plot_throughput_analysis(stats_csv)
```
